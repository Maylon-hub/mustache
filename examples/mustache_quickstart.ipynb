# MustaCHE quick start

This notebook verifies the public Python API with a small reproducible batch analysis.

```python
import pandas as pd
from sklearn.datasets import make_blobs
import mustache
from mustache.core.batch import run_batch_clustering

print(f'MustaCHE {mustache.__version__}')
```

```python
X, _ = make_blobs(n_samples=90, n_features=3, centers=3, random_state=42)
data = pd.DataFrame(X, columns=['x', 'y', 'z'])
results = run_batch_clustering(
    data, min_mpts=4, max_mpts=8, step=2, metric='euclidean', algorithm='core-sg'
)
assert list(results) == ['4', '6', '8']
assert all(len(item['labels']) == len(data) for item in results.values())
print(f'Processed {len(results)} hierarchies for {len(data)} samples.')
```
